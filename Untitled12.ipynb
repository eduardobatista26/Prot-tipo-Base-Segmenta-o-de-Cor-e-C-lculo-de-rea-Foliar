```python
import cv2
import numpy as np

# 1. Carregar a imagem da muda
img = cv2.imread('muda_planta.jpg')

# Conversão do espaço de cor BGR (padrão OpenCV) para HSV
# O espaço HSV facilita o isolamento de cores específicas independentemente da iluminação
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# 2. Definir o intervalo da cor verde (Ajustável conforme o sensor/iluminação)
verde_baixo = np.array([35, 40, 40])
verde_alto = np.array([85, 255, 255])

# 3. Criar a máscara binária (pixels verdes ficam brancos=255, o resto fica preto=0)
mascara = cv2.inRange(hsv, verde_baixo, verde_alto)

# 4. Contar a quantidade total de pixels brancos (área da planta)
pixels_planta = cv2.countNonZero(mascara)

# 5. Conversão Espacial (Exemplo: escala onde 100 pixels equivalem a 1 cm²)
fator_escala = 100  # pixels/cm²
area_cm2 = pixels_planta / fator_escala

print(f"Área Foliar Estimada: {area_cm2:.2f} cm²")

# Exibir a máscara resultante da segmentação
cv2.imshow('Planta Isolada (Máscara Binária)', mascara)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
